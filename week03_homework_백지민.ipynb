### Problem 19: Mini Vocabulary Quiz (one question)

Create two variables: `question_word` (e.g. `"apple"`) and `correct_meaning` (e.g. `"사과"`). Ask the user to type the meaning using `input()`. Write a function `check_answer(user_answer, correct_answer)` that returns `True` or `False` (ignore uppercase/lowercase and extra spaces using `.strip().lower()`). Print `"Correct!"` if the function returns `True`, otherwise print `"Try again!"`.

💡 **Hint — combining a function with `if`:**
```python
def is_even(number):
    return number % 2 == 0

n = 4
if is_even(n):
    print("Even!")
else:
    print("Odd!")
```


````python
question_word = "beef"
correct_meaning = "소고기"

user_answer = input(f"Enter the meaning of {question_word}: ")


def check_answer(user_answer, correct_answer):
    if user_answer.strip().lower() == correct_answer.strip().lower():
        return True
    else:
        return False


if check_answer(user_answer, correct_meaning):
    print("Correct!")
else:
    print("Try again!")
````

````python
# TODO: write your code here
import random

word_list = ["cat", "dog", "bird", "lion", "tiger"]
meaning_list = ["고양이", "개", "새", "사자", "호랑이"]

def check_answer(user_answer, correct_answer):
    return user_answer.strip().lower() == correct_answer.strip().lower()

score = 0
total_questions = 3

for i in range(total_questions):
    index = random.randint(0, len(word_list) - 1)
    user_answer = input(f"Enter the meaning of {word_list[index]}: ")

    if check_answer(user_answer, meaning_list[index]):
        print("Correct!")
        score += 1
    else:
        print("Try again!")
    pass
print(f"Final score: {score}/{total_questions}")
````
